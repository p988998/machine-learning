```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the function y = sin(x)
x = np.arange(0, 6, 0.1)
y = np.sin(x)
```

```python
print(x)
```

```python
print(y)
```

```python
# Plot the function graph
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sin(x)")
plt.show()
```

```python
# Derivative y' = cos(x)
y1 = np.cos(x)
print(y1)
```

```python
# Plot the derivative graph
plt.plot(x, y1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = cos(x)")
plt.show()
```

```python
# Plot both the function and its derivative on the same figure
plt.plot(x, y, label="y=sin(x)")
plt.plot(x, y1, label="y'=cos(x)", linestyle="--")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sin(x)")
plt.legend()
plt.show()
```
